These line graphs shows number of trips where people with mobility has travelled in the different purposes e.g., 'Business, Shopping and Education'.

```python
sns.lineplot(data=mobility,x='Year',y='Business',errorbar=None,hue='Mobility status ',marker='o')
plt.title('Mobility Business Trips 2007-2025')
plt.ylabel('Average Number of Trips')
plt.show()
```

```python
impairment['Trips per person per year'] = impairment['Trips per person per year'].replace({',':''},regex=True).astype(int)
```

```python
sns.barplot(data=impairment,x='Year',y='Trips per person per year',hue='Sex ',palette='Set1',errorbar=None)
plt.title('Impairment Trips per gender 2018-2025')
plt.legend(loc='lower left')
plt.show()
```

```python
sns.barplot(data=impairment,x='Impairment ',y='Trips per person per year',hue='Impairment ',palette='Set1',errorbar=None)
plt.title('Impairment Trips')
plt.xlabel('Impairment types')
plt.xticks(rotation=87)
plt.show()
```
