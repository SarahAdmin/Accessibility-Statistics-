```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
employment = pd.read_csv('Disability work status.csv')
```

```python
licence = pd.read_csv('driving_licence_data.csv')
```

```python
employment.columns
```

```python
employment['Full time'] = employment['Full time'].replace({',':''},regex=True).astype(int)
```

```python
employment['Part time'] = employment['Part time'].replace({',':''},regex=True).astype(int)
```

```python
employment['Retired or permanently sick'] = employment['Retired or permanently sick'].replace({',':''},regex=True).astype(int)
```

```python
employment['Other non-work '] = employment['Other non-work '].replace({',':''},regex=True).astype(int)
```

```python
employment['All people'] = employment['All people'].replace({',':''},regex=True).astype(int)
```

```python
employment.drop(columns='Retired or permanetly sick')
```

```python
employment.info()
```

```python
sns.barplot(data=employment,x='Year',y='Full time',hue='Disability status',palette='bright',errorbar=None)
plt.title('Full Time Employment via Disability status 2018-2025')
plt.ylabel('Average number of Trips')
plt.legend(loc='lower right')
plt.show()
```

```python
sns.barplot(data=employment,x='Year',y='Part time',hue='Disability status',palette='bright',errorbar=None)
plt.title('Part Time Employment via Disability Status 2018-2025')
plt.ylabel('Average number of Trips')
plt.legend(loc='lower right')
plt.show()
```

```python
sns.barplot(data=employment,x='Year',y='Retired or permanently sick',hue='Disability status',palette='bright',errorbar=None)
plt.title('Retired or Permanetly Sick via Disability Status 2018-2025')
plt.ylabel('Average number of Trips')
plt.legend(loc='lower right')
plt.show()
```

```python
sns.barplot(data=employment,x='Year',y='Other non-work ',hue='Disability status',palette='bright',errorbar=None)
plt.title('Other Non-Employment via Disability Status 2018-2025')
plt.ylabel('Average number of Trips')
plt.legend(loc='lower right')
plt.show()
```

```python
sns.barplot(data=employment,x='Year',y='Full time',hue='Age group',palette='Set1',errorbar=None)
plt.title('Full Time Employment via Age groups 2018-2025')
plt.ylabel('Average number of Trips')
plt.legend(loc='lower right')
plt.show()
```

```python
sns.barplot(data=employment,x='Year',y='Part time',hue='Age group',palette='Set1',errorbar=None)
plt.title('Part Time Employment via Age groups 2018-2025')
plt.ylabel('Average number of Trips')
plt.legend(loc='lower right')
plt.show()
```

```python
sns.barplot(data=employment,x='Year',y='Retired or permanently sick',hue='Age group',palette='Set1',errorbar=None)
plt.title('Retired or Permanently Sick via Age groups 2018-2025')
plt.ylabel('Average number of Trips')
plt.legend(loc='lower right')
plt.show()
```

```python
sns.barplot(data=employment,x='Year',y='Other non-work ',hue='Age group',palette='Set1',errorbar=None)
plt.title('Other Non-Employment via Age groups 2018-2025')
plt.ylabel('Average number of Trips')
plt.legend(loc='lower right')
plt.show()
```
