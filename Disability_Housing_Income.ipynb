```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
housing = pd.read_csv('housing income.csv')
```

```python
housing.head(10)
```

```python
housing.info()
```

```python
housing['Second level'] = housing['Second level'].replace({',':''},regex=True).astype(int)
```

```python
housing['Second level'] = housing['Second level'].replace({',':''},regex=True).astype(int)
```

```python
housing['Third level'] = housing['Third level'].replace({',':''},regex=True).astype(int)
```

```python
housing['Fourth level'] = housing['Fourth level'].replace({',':''},regex=True).astype(int)
```

```python
housing['Highest level'] = housing['Highest level'].replace({',':''},regex=True).astype(int)
```

```python
housing.dtypes
```

```python
fig, ax = plt.subplots(figsize=(10,5))
sns.lineplot(data=housing,x='Year',y='Lowest level',hue='Disability status',marker='o',palette='Set1')
plt.title('Lowest Housing Income Quartile by Disability Status 2018-2025')
plt.ylabel('Average Number of Trips')
for x,y in zip(housing['Year'],housing['Lowest level']):
  ax.text(x,y,str(y),
          horizontalalignment='center',
          fontsize=10,
          fontweight='normal')
plt.show()
```
